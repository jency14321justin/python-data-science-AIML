```python
#multiple inheritance
class father:
    def skills(self):
        return "painting"
class mother:
     def skills(self):
         return "cooking"
class child(father,mother):
    def skills(self):
        return father.skills(self) + "and" + mother.skills(self)
c=child()
print(c.skills())



class construction:
    def workers(self):
        return "starting the work"
class status:
    def workers(self):
        return " work is almost finish"
class buildings(construction,status):
    def workers(self):
        return  construction.workers(self) + "and" + status.workers(self)
b=buildings()
print(b.workers())

class mobiles:
    def shops(self):
        return "new mobiles are avaliable"
class dresses:
    def shops(self):
        return "new models are arrived"
class mall(mobiles,dresses):
    def shops(self):
        return mobiles.shops(self) + "and" + dresses.shops(self)
m=mall()
print(m.shops())    

class books:
    def shops(self):
        return "choose books "
class jucies:
    def shops(self):
        return "choose jucies"
class margin_free:
    def shops(self):
        return books.shops(self) + "or" + jucies.shops(self)

m=margin_free()
print(m.shops())
```

```python
#abstract method
from abc import ABC,abstractmethod
class shape(ABC):
    @abstractmethod
    def area(self):
        pass
    def peri(self):
        pass
class circle(shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return 3.14 * self.radius *self.radius
class rectangle(shape):
    def __init__(self,l,b):
        self.l=l
        self.b=b
    def area(self):
        return  self.l * self.b
c=circle(5)
print('area of the circle ',c.area())
r=rectangle(5,5)
print('area of the rectangle',r.area())


from abc import ABC,abstractmethod 
class bankbalance(ABC):
    @abstractmethod 
    def account(self):
        pass
    def balance(self):
        pass
class withdraw (bankbalance):
    def __init__(self,withdraw):
        self.withdraw =withdraw
    def account(self):
            return self.withdraw 
class deposit(bankbalance):
    def __init__(self,deposit):
        self.deposit=deposit
    def account(self):
        return self.deposit
print(int(input("enter pin")))
print(int(input("enter the balance :")))
a=withdraw(100)
print('withdraw successfully',a.account())
b=deposit(500)
print('deposit successfully',b.account())
    
    
    
```

```python
#Hierarchical Inheritance
class parent:
    def property(self):
        return "has land property"
class son(parent):
    pass
class daughter(parent):
    pass
s=son()
d=daughter()
print(s.property())
print(d.property())


class bakery:
    def cake(self):
        return "Has in bakery."
class chips(bakery):
    pass
class biscuits(bakery):
    pass
class sweets(bakery):
    pass
c=chips()
b=biscuits()
s=sweets()
print(c.cake())
print(b.cake())
print(s.cake())



```

```python
#### Hybrid Inheritance
class A:
    def method_A(self):
        return "method from A"
class B(A):
    def method_B(self):
        return "method from B"
class C(A):
    def method_C(self):
        return "method from C"
class D(C,D):
    def method_D(self):
        return "method from D"
obj=D()
print(obj.method_A())
print(obj.method_B())
print(obj.method_C())
print(obj.method_D())
    
```
