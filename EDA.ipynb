```python
#categorical encodind 
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['species']=le.fit_transform(df['species'])
unique_species =df['species'].unique()
print("Encoder species values :")
print(unique_species)
```

```python
#univariate analysis
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
sns.histplot(iris['sepal_length'],kde=True,ax=axes[0,0],color='brown')
axes[0,0].set_title('sepal length Distribution')

sns.histplot(iris['sepal_width'],kde=True,ax=axes[0,1],color='black')
axes[0,1].set_title('sepal width Distribution')

sns.histplot(iris['petal_length'],kde=True,ax=axes[1,0],color='green')
axes[1,0].set_title('petal length Distribution')

sns.histplot(iris['petal_width'],kde=True,ax=axes[1,1],color='grey')
axes[1,1].set_title('petal width Distribution')
plt.tight_layout()
plt.show()
```

```python
#boxplots for numerical features
plt.figure(figsize=(12,6))
sns.boxplot(data=iris.drop('species',axis=1))
plt.title('boxplot of all numerical features')
plt.ylabel('centimeters(cm)')
plt.grid(True,alpha=1,color='black')
plt.show()
```

```python
df=pd.DataFrame(iris)
df["species"].value_counts().plot.pie(explode=[0.1,0.1,0.1],autopct='%1.1f%%'
,shadow=True, figsize=(10,8))
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.suptitle('Distribution of Iris Flower Measurement',fontsize=14)
plt.subplot(2,2,1)
plt.hist(df['sepal_length'])
plt.title('sepal length diatribution')
plt.grid(True,alpha=1,color='green')

plt.subplot(2,2,2)
plt.hist(df['sepal_width'])
plt.title('sepal width distribution')
plt.grid(True,alpha=0.30,color='black')

plt.subplot(2,2,3)
plt.hist(df['petal_length'])
plt.title('petal length distribution')
plt.grid(True,alpha=1,color='black')

plt.subplot(2,2,4)
plt.hist(df['petal_width'])
plt.title('petal width distribution')
plt.grid(True,alpha=1,color='black')
```

```python
plt.figure(figsize=(8, 6))
species=df['species'].unique()
colors = ['red', 'green', 'blue']
for i in range(len(species)):
    x = df[df['species'] == Species[i]]
    plt.scatter(x['sepal_length'], x['sepal_width'], c=colors[i], label=species[i])

plt.xlabel('sepal length (cm)')
plt.ylabel('sepal width (cm)')
plt.title('sepal length vs width by species')
plt.legend()
plt.show()
```
