```python
#create database
import sqlite3
conn=sqlite3.connect("my_database.df")
#create cursor object
cursor=conn.cursor()
#creatin database table
cursor.execute("""
CREATE TABLE IF NOT EXISTS students(
    id INT,
    name TEXT,
    age INT, 
    mark INT
)
""")
print ("table is created successfully")

#insert the values in the table 
cursor.execute("INSERT INTO students (id,name, age, mark) VALUES (?, ?, ?,?)",
(1,"Alice", 21, 75))

# insert multiple table 
students = [
(2,"Bob", 22, 56),    # inside the list we get value as tuple 
(3,"Charlie", 20, 53),
(4,"Diana", 23, 85)
]
cursor.executemany(
    "INSERT INTO students (id,name, age, mark) VALUES (?, ?, ?,?)", students)
 #con.commit uses to save the table 
conn.commit()  
print("Data inserted successfully!")

#selecting students
cursor.execute("SELECT* FROM students")
row=cursor.fetchall()
print("all students")
for row in row:
    print(row)


    
```
