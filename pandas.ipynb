```python
#data cleaning
import pandas as pd 
data={"Name":["jency","joyeelin","pinky",None,"mano","akshay","ugin"],
      "age":[21,25,25,None,29,28,30],
      "salary":[45000,50000,None,46000,12000,45000,90000],
      "department":["HR","AI",None,"EEE","HR","ML","HM"]}
employees=pd.DataFrame(data)
print("original data")
display(employees)

print("\n missing value:")
display(employees.isna())
display(employees.isna().sum())

clean_df=employees.copy()
clean_df["age"]=clean_df["age"].fillna(clean_df["age"].mean())
display(clean_df)

clean_df.dropna(subset=["Name","department"],inplace= True)
clean_df.drop_duplicates(inplace=True)
print("\ncleaned data:")
display(clean_df)
```

```python
#DATA MANIPULATION
import pandas as pd
import numpy as np
# create sample dataframe
sales=pd.DataFrame({
    "Date":pd.date_range("2026-10-13",periods=10),
    "product":np.random.choice(["Widget", "Gadget", "Tool"],10),
    "units":np.random.randint(1,20,10),
    "unit_price":np.random.uniform(10,50,10).round(2)
        })
sales
```

```python
sales["Total"]=sales["units"]*sales["unit_price"]
print("sales data with total")
display(sales)
```

```python
#applying lambda function
sales["price_category"]=sales["unit_price"].apply(
    lambda x :"HIGH" if x>25 else "LOW")
print("\n with price_category")
display(sales)
```

```python
#binning
sales["unit_Bin"]=pd.cut(
    sales["units"],
    bins=[0,5,10,20,30],
    labels=["LOW","MEDIUM" ,"HIGH","VERY HIGH"],
    include_lowest=True
)
print("\n with binned units")
display(sales)

```
