```python
#data cleaning
import pandas as pd

df=pd.DataFrame(titanic_df)
df=df[['survived','pclass','sex','age','sibsp','parch','fare','embarked']]
#titanic_df['sex']=titanic_df['sex']

display(df)

```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['sex'] = le.fit_transform(df['sex'])
df
```

```python
df['embarked'] = le.fit_transform(df['embarked'])
df
```

```python
df["age"]=df["age"].fillna(df["age"].mean())
df
```

```python
#univarient
fig,axes= plt.subplots(4,2, figsize=(18,20))
sns.histplot(titanic_df['survived'],kde=True,ax=axes[0,0],color='black')
axes[0,0].set_title('Titanic survived Distribution')

sns.histplot(titanic_df['pclass'],kde=True,ax=axes[0,1],color='red')
axes[0,1].set_title('Titanic pclass Distribution')

sns.histplot(titanic_df['sex'],kde=True,ax=axes[1,0],color='orange')
axes[1,0].set_title('Titanic sex distribution')

sns.histplot(titanic_df['age'],kde=True,ax=axes[1,1],color='blue')
axes[1,1].set_title('Titanic age distribution')

sns.histplot(titanic_df['sibsp'],kde=True,ax=axes[2,0],color='green')
axes[2,0].set_title('Titanic sibsp distribution')

sns.histplot(titanic_df['parch'],kde=True,ax=axes[2,1],color='yellow')
axes[2,1].set_title('Titanic parch distribution')

sns.histplot(titanic_df['fare'],kde=True,ax=axes[3,0],color='skyblue')
axes[3,0].set_title('Titanic fare distribution')



sns.histplot(titanic_df['embarked'],kde=True,ax=axes[3,1],color='pink')
axes[3,1].set_title('Titanic embarked distribution')



plt.show()
```

```python
#boxplots for numerical features
plt.figure(figsize=(10,6))
sns.boxplot(data=titanic_df.drop('embarked',axis=1))
plt.title('boxplot of all numerical features')
plt.ylabel('centimeters(cm)')
plt.show()            
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
titanic_df=sns.load_dataset('titanic')
titanic_df['survived'].value_counts().plot.pie(
    explode=[0.1,0.1],
    autopct='%1.1f%%',
    shadow=True,
    figsize=(10,8))
plt.show()
```
