```python
#Titanic Dataset Attribute Descriptions
import pandas as pd
titanic_df=pd.read_csv("titanic.csv")
display(titanic_df)

```

```python
embarked=titanic_df["Embarked"].value_counts()
print(embarked.index)
print(embarked.values)
display(embarked)
plt.bar(x=embarked.index , height=embarked.values,alpha=0.5,color=['red','blue','yellow'])
```

```python
embarked_m = titanic_df[titanic_df["Gender"]=='male']["Embarked"].value_counts()
embarked_f = titanic_df[titanic_df["Gender"]=='female']["Embarked"].value_counts()
print(embarked_m)
print(embarked_f)

plt.bar(x=embarked_f.index,height=embarked_f.values,color="lavender")
plt.bar(x=embarked_m.index,height=embarked_m.values,bottom=embarked_f,color="red")
plt.legend(labels=['M','F'])
plt.grid(True,alpha=1,color='blue')
plt.title('port of embarkation by  sex')
plt.xlabel('port')
plt.ylabel('passangers')
```

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(start=-np.pi,stop=np.pi,num=256)
y=np.sin(2*x)
fig=plt.figure(figsize=(10,8))
plt.plot(x,y)
plt.text(-1.65,0.5,'y=sin(2x)',fontsize=14);
```

```python
fig=plt.figure(figsize=(10,8))
plt.plot(x,y, color='pink')
plt.plot([0],[0],'o')
arrowprops=dict(facecolor='black',shrink=0.1,width=2)
plt.annotate('origin point',
             xy=(0, 0),
             xytext=(-1.65,0.5),
             arrowprops=arrowprops,
             fontsize=12);
          
```

```python
#3d plots
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt

age=titanic_df['Age']
fare=titanic_df['Fare']
pclass=titanic_df['Pclass']

fig=plt.figure(figsize=(9,9))
ax=plt.axes(projection='3d')
ax.scatter(xs=age,ys=fare,zs=pclass)

ax.set_xlabel('Age')
ax.set_ylabel('Fare')
ax.set_zlabel('Pclass') 
plt.show()
```

```python
import matplotlib.pyplot as plt
s_vals = titanic_df['Gender'].value_counts()
e_vals = titanic_df['Embarked'].value_counts()
fig, axes = plt.subplots(1, 2)
fig.set_size_inches(8, 4) 
axes[0].pie(s_vals, labels=['Male', 'Female'], startangle=90, autopct='%.1f%%')
axes[0].set_title('Gender')

axes[1].pie(e_vals, labels=['S', 'C', 'Q'], startangle=90, explode=[0, 0.1, 0])
axes[1].set_title('Embarked');
```
